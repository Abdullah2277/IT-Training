### Class Task 1

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def linearFunction(x,m = 2,c = 3):
  return m*x + c
```

```python
x = np.linspace(-10, 10, 100)

y = linearFunction(x)
```

```python

plt.plot(x, y, label='y = 2x+3')
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)
plt.show()
```

### Class Task 2

```python
def gradient_descent(x, y, learning_rate, iterations):
    m = 0
    c = 0
    n = len(x)

    for i in range(iterations):
        y_predicted = m * x + c
        dm = (-2 / n) * sum(x * (y - y_predicted))
        dc = (-2 / n) * sum(y - y_predicted)
        m = m - learning_rate * dm
        c = c - learning_rate * dc

    return m, c

x = np.array([1, 2, 3, 4, 5])
y = np.array([3, 5, 7.5, 8, 11])

learning_rate = 0.01
iterations = 500

m, c = gradient_descent(x, y, learning_rate, iterations)

print("Optimized m:", m)
print("Optimized c:", c)
```
